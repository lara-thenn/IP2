**Theory:**

Het (I,V)-karakteristiek kan worden beschreven als:

\begin{equation}
    I_D(V_D) = I_0 \left( e^{\frac{-q V_D}{n k_B T}} - 1 \right)
    \label{eq:diode}
\end{equation}

Hierbij is $I_{D}$ de stroom door de diode, $V_{D}$ de spanning over de diode, $I_{0}$ de sperstroom, $q$ is de lading van een elektron, $n$ is de idealiteitsfactor, $T$ is de temperatuur uitgedrukt in Kelvin, en $k_{B}$ is de constante van Boltzmann.
Uit de gemeten bronspanning $U_\mathrm{bron}$ en de spanning over de serieweerstand $U_{R_1}$ zijn de diodestroom en de diodespanning berekend:
\begin{equation}
    I_D = \frac{U_{R_1}}{R_1}
    \label{eq:I_D}
\end{equation}
en
\begin{equation}
    U_D = \frac{R_3}{R_2 + R_3}\, U_\mathrm{bron} - U_{R_1}.
    \label{eq:U_D}
\end{equation}

De onzekerheden in $U_\mathrm{bron}$ en $U_{R_1}$ zijn bepaald uit de specificaties van de multimeters (een percentage van de aflezing plus een aantal counts) en omgerekend naar een standaardonzekerheid door te delen door $\sqrt{3}$.

De data is weergegeven als $I_D$ tegen $U_D$ op een logaritmische schaal, naar het voorbeeld van de (I,V)-karakteristiek. Om goede startwaarden voor de fit te vinden, is eerst een rechte lijn gefit door $\log_{10}(I_D)$ voor de punten met $U_D > 0{,}03\,\mathrm{V}$. Daarna is met curve\_fit de volledige Shockley-vergelijking
\begin{equation}
    I_D = I_s \left( e^{a U_D} - 1 \right)
    \label{eq:shockley}
\end{equation}
gefit, gewogen met de onzekerheid in $I_D$.

Uit de fitparameter $a = \frac{q}{n k_B T}$ volgt
\begin{equation}
    k_B = \frac{q}{a\, n\, T},
    \label{eq:kB}
\end{equation}
met $n = 2{,}0 \pm 0{,}1$ en $T$ de gemiddelde gemeten temperatuur. De onzekerheid in $k_B$ is bepaald door de relatieve onzekerheden in $a$, $n$ en $T$ kwadratisch op te tellen:
\begin{equation}
    \frac{u(k_B)}{k_B} = \sqrt{\left(\frac{u(a)}{a}\right)^2 + \left(\frac{u(n)}{n}\right)^2 + \left(\frac{u(T)}{T}\right)^2}.
    \label{eq:ukB}
\end{equation}

# Execution

```python
from scipy.constants import e as q

def shockley(V, Is, a):
    return Is * (np.exp(a * V) - 1)

# Deze functie zorgt ervoor dat we de juiste startwaarde vinden voor de fit
# van de shockley functie.
def lijn(x, c, d):
    return c * x + d

R = 10000
R2 = 470
R3 = 10
n = 2.0
sn = 0.1

# Gemeten data.
temp_avg = np.array([24.904, 24.952, 24.956, 24.954, 24.933, 24.946, 24.972, 24.980, 24.954, 24.986])
USB = np.array([6.02, 5.77, 5.52, 5.26, 5.02, 4.74, 4.54, 4.24, 3.99, 3.74, 3.49, 3.24, 2.99, 2.74, 2.49, 2.24, 1.99, 1.74, 1.49, 1.24, 0.99, 0.74, 0.49, 0.24])
VR1 = np.array([0.417, 0.323, 0.296, 0.259, 0.233, 0.241, 0.198, 0.194, 0.153, 0.124, 0.116, 0.104, 0.091, 0.074, 0.063, 0.051, 0.042, 0.036, 0.028, 0.021, 0.016, 0.010, 0.007, 0.002]) / 1e3

# De temperatuur, met onzekerheid, in Kelvin.
T  = np.mean(temp_avg) + 273.15
sT = 1.0
I_D = VR1 / R
U = R3 / (R2 + R3) * USB
U_D = U - VR1

# Onzekerheden in de meetwaarden.
sR   = 0.01 * R / np.sqrt(3)
sUSB = (0.0050 * USB + 0.01) / np.sqrt(3)
sVR1 = (0.00015 * VR1 + (0.0003 / 1e3)) / np.sqrt(3)

# Onzekerheid in de diodestroom voor de fit (I_D = VR1 / R).
sI_D = sVR1 / R

# Voor het vinden van die startwaarde gebruiken we een fit met een rechte lijn, wat we in principe verwachten van de grafiek.
# We willen alleen punten die boven de 0.03 uitkomen, I_d > 0 omdat je niet de log van <= 0 kan nemen.
# p0 rekent de parameters om naar waarden die passen bij de shockley functie met a = c *ln(10) = c/log10(e)

mask = (U_D > 0.03) & (I_D > 0)
(c, d), _ = curve_fit(lijn, U_D[mask], np.log10(I_D[mask]))
p0 = [10 ** d, c / np.log10(np.e)]

# De daadwerkelijke fit op de Shockley functie, gewogen met de onzekerheid in I_D.
popt, pcov = curve_fit(shockley, U_D, I_D, p0=p0, sigma=sI_D, absolute_sigma=False)
Is, a = popt
sIs, sa = np.sqrt(np.diag(pcov))

# Constante van Boltzmann met onzekerheid
kB  = q / (a * n * T)
skB = kB * np.sqrt((sa / a)**2 + (sn / n)**2 + (sT / T)**2)

# De waarde van de constante van Boltzmann zoals wij deze berekend hebben.
# Daarnaast ook de meetwaarden met de onzekerheden.
print(f"Is  = {Is:.3e} ± {sIs:.1e} A")
print(f"a   = {a:.3g} ± {sa:.2g} 1/V")
print(f"k_B = {kB:.3e} ± {skB:.1e} J/K   (literatuur: 1.381e-23)")
print("Meetwaarden met onzekerheden:")
print(f"{'#':>3} {'USB (V)':>16} {'VR1 (mV)':>20}")
print(T)
for i, (u, su, v, sv) in enumerate(zip(USB, sUSB, VR1, sVR1), start=1):
    print(f"{i:>3} {u:7.2f} +- {su:.3f} {v*1e3:8.3f} +- {sv*1e3:.6f}")

# De bijbehorende grafieken.
V = np.linspace(U_D.min(), U_D.max(), 500)
plt.errorbar(U_D, I_D, yerr=sI_D, fmt='o', label='meting')
plt.semilogy(V, shockley(V, *popt), label='fit')
plt.yscale('log')
plt.xlabel('$U_D$ (V)')
plt.ylabel('$I_D$ (A)')
plt.legend()
plt.savefig('Plot gemeten waarden.png')
plt.show()
```

**Observations:** 

De plot volgt hetzelfde patroon als de gegeven grafiek in de opdracht. De berekende constante van Boltzmann verschilt heel weinig van de echte waarde.
